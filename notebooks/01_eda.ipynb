# Load the synthetic data and check the shape of each domain

```python
import pandas as pd
mig = pd.read_csv('data/processed/migration.csv', parse_dates=['migration_date'])
hl7 = pd.read_csv('data/processed/hl7_messages.csv', parse_dates=['msg_datetime'])
fhir = pd.read_csv('data/processed/fhir_resources.csv', parse_dates=['created_date'])
tmap = pd.read_csv('data/processed/terminology_map.csv')
print(mig.shape, hl7.shape, fhir.shape, tmap.shape)
print(mig['break_type'].value_counts(normalize=True).round(4))
print(hl7['defect_type'].value_counts(normalize=True).round(4))
print(tmap.groupby('target_system')['mapped'].mean().round(3))
```

# Break rate by migration wave — does the go-live wave get worse?

```python
print(mig.groupby('migration_wave')['has_break'].mean().round(4))
```

# Conformance by source system and standard

```python
iface = pd.read_csv('data/processed/interface_combined.csv')
print(iface.groupby(['standard', 'source_system'])['nonconformant'].mean().round(4))
```
